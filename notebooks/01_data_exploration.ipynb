```python
%sql
SELECT
COUNT(*) AS total_rows,
COUNT(match_id) AS match_id_not_null,
COUNT(*) - COUNT(match_id) AS match_id_nulls
FROM matches;
```

```python
%sql
SELECT
COUNT(*) AS total_rows,
COUNT(delivery_id) AS delivery_id_not_null,
COUNT(*) - COUNT(delivery_id) AS delivery_id_nulls
FROM deliveries;
```

```python
%sql
SELECT
COUNT(*) AS total_rows,
COUNT(player_id) AS player_id_not_null,
COUNT(*) - COUNT(player_id) AS player_id_nulls
FROM players;
```

```python
%sql
SELECT
COUNT(*) AS total_rows,
COUNT(venue_id) AS venue_id_not_null,
COUNT(*) - COUNT(venue_id) AS venue_id_nulls
FROM venues;
```

```python
%sql
SELECT COUNT(*) AS total, COUNT(DISTINCT match_id) AS unique_match_ids
FROM matches;
```

```python
%sql
SELECT COUNT(*) AS total, COUNT(DISTINCT delivery_id) AS unique_delivery_ids
FROM deliveries;
```

```python
%sql
SELECT COUNT(*) AS total, COUNT(DISTINCT player_id) AS unique_player_ids
FROM players;
```

```python
%sql
SELECT COUNT(*) AS total, COUNT(DISTINCT venue_id) AS unique_venue_ids
FROM venues;
```

```python
%sql
SELECT
m.match_id,
COUNT(d.delivery_id) AS deliveries
FROM matches m
JOIN deliveries d
ON m.match_id = d.match_id
GROUP BY m.match_id
LIMIT 10;
```

```python
%sql
SELECT
m.match_id,
v.venue_name
FROM matches m
JOIN venues v
ON m.venue_id = v.venue_id
LIMIT 10;
```

```python
%sql
SELECT
d.delivery_id,
p.player_name
FROM deliveries d
JOIN players p
ON d.batter_id = p.player_id
LIMIT 10;
```

```python
%sql
SELECT
COUNT(*) AS total_rows,
COUNT(DISTINCT m.match_id) AS unique_matches
FROM matches m
JOIN deliveries d
ON m.match_id = d.match_id;
```

```python
%sql
SELECT match_id, COUNT(*)
FROM matches
GROUP BY match_id
HAVING COUNT(*) > 1;
```

```python
%sql
SELECT
match_id,
innings_no,
over_no,
ball_seq,
COUNT(*)
FROM deliveries
GROUP BY
match_id,
innings_no,
over_no,
ball_seq
HAVING COUNT(*) > 1;
```

```python
%sql
SELECT player_id, COUNT(*)
FROM players
GROUP BY player_id
HAVING COUNT(*) > 1;
```

```python
%sql
SELECT venue_id, COUNT(*)
FROM venues
GROUP BY venue_id
HAVING COUNT(*) > 1;
```
